```python
# import pandas as pd
# import numpy as np

# Set random seed
np.random.seed(0)

# Number of students
n = 200

# Generate student data
hours_studied = np.random.randint(1, 10, n)

attendance = np.random.randint(60, 101, n)

assignments_completed = np.random.randint(5, 11, n)

previous_score = np.random.randint(40, 91, n)

# Calculate Final Score
final_score = (
    0.30 * (hours_studied * 10)
    + 0.20 * attendance
    + 0.20 * (assignments_completed * 10)
    + 0.30 * previous_score
)

# Round the final score
final_score = final_score.round(2)

# Create DataFrame
dfff = pd.DataFrame({
    "Hours_Studied": hours_studied,
    "Attendance": attendance,
    "Assignments_Completed": assignments_completed,
    "Previous_Score": previous_score,
    "Final_Score": final_score
})

# Display first 10 rows
print(dfff.head(10))

# Check dataset shape
print("\nDataset Shape:", dfff.shape)

# Check for missing values
print("\nMissing Values:")
print(dfff.isnull().sum())

# Check for duplicates
print("\nDuplicate Rows:", dfff.duplicated().sum())

# Save as CSV
dfff.to_csv("study_scores_multiple_linear_regression.csv", index=False)

print("\nCSV file saved successfully!")
```
